# Exploring octal games

The C backend (`octal.c`) computes nim-sequences; `octalgames` wraps it for Python. Codes look like `.007` (Treblecross), `.6` (Officers), `.137` (Dawson's chess), `4.` (split a heap), or the names `grundy` and `couples`.

```python
%matplotlib inline
import octalgames as og
import numpy as np
```

## A solved game: Kayles (.77)
The period 12 is *proved* by the octal periodicity theorem.

```python
kayles = og.compute('.77', 2000)
kayles
```

```python
kayles.plot_wrap(mode='values', rows=60);
```

In `diff` mode a cell is dark when G(n) ≠ G(n−W). With W = 12, everything after the preperiod (n = 71) is blank.

```python
kayles.plot_wrap(12, mode='diff', rows=60);
```

## An open problem: Treblecross (.007)
No period is known (Nowakowski, Problem A2). Interrupt the kernel at any time; computed values are kept, and `.extend(n)` continues later.

```python
t = og.compute('.007', 100_000)
t
```

```python
t.dashboard();
```

```python
og.plot_first_occurrence(t);
```

## Interactive explorer
Change the width W with the slider or the buttons. When W hits a period, the picture turns into stripes. Try `diff` mode, the bit planes, and `values ≥ threshold` (set *bit / thr* to e.g. 256).

```python
t.explore()
```

## Survey: all two-digit octal games
This computes 2,000 values each and reports proved or apparent periods.

```python
df = og.survey(og.octal_codes(2), 2000)
df[['period', 'preperiod', 'proven', 'max_value']].sort_values('period', na_position='first').head(20)
```

```python
df[df.period.isna()].index.tolist()   # no period visible within 2000 values
```
